```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Function to read data from CSV files
def read_csv_data(file_name):
    data = pd.read_csv(file_name)
    actual_areas = data['Actual_Area'].values
    predicted_areas = data['Predicted_Area'].values
    return actual_areas, predicted_areas

# Function to load and preprocess data from multiple CSV files
def load_data(file_names):
    actual_areas_list = []
    predicted_areas_list = []

    for file_name in file_names:
        actual_areas, predicted_areas = read_csv_data(file_name)
        actual_areas_list.append(actual_areas)
        predicted_areas_list.append(predicted_areas)

    return np.concatenate(actual_areas_list), np.concatenate(predicted_areas_list)

# List of CSV files
file_names = [f"area_data_{i}.csv" for i in range(1, 51)]  # Assuming files are named as area_data_1.csv, area_data_2.csv, ...

# Load and preprocess data
actual_areas, predicted_areas = load_data(file_names)

# Normalize features if needed
scaler = StandardScaler()
actual_areas_normalized = scaler.fit_transform(actual_areas.reshape(-1, 1))

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(actual_areas_normalized, predicted_areas, test_size=0.2, random_state=42)

# Define the neural network architecture
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(1,)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(1)
])

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mse'])

# Train the model
history = model.fit(X_train, y_train, epochs=50, batch_size=32, validation_split=0.2)

# Evaluate the model
test_loss, test_mse = model.evaluate(X_test, y_test)
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
print(f'Test Loss: {test_loss}, Test MSE: {test_mse}, Test MAE: {mae}, Test RMSE: {np.sqrt(mse)}')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the target distribution (random walk)
def target_distribution(x, mean):
    return np.exp(-0.5 * ((x - mean)**2))

# Metropolis-Hastings MCMC algorithm for mean estimation
def metropolis_hastings_mean_random_walk(iterations, proposal_sd, true_walk):
    predictions = []
    current_mean = 0

    for i in range(len(true_walk)):
        # Propose a new observation from a normal distribution centered at the current mean
        observation = current_mean + np.random.normal(0, proposal_sd)

        # Calculate the acceptance ratio
        acceptance_ratio = min(1, target_distribution(observation, current_mean) / target_distribution(true_walk[i], current_mean))

        # Accept or reject the observation based on the acceptance ratio
        if np.random.rand() < acceptance_ratio:
            predictions.append(observation)
            current_mean = np.mean(predictions)

    # Pad predictions with zeros to match the length of true_walk
    predictions += [0] * (len(true_walk) - len(predictions))
    return predictions

if __name__ == "__main__":
    np.random.seed(42)  # for reproducibility
    num_points = 1000
    true_mean = 0
    true_walk = np.cumsum(np.random.normal(0, 1, num_points)) + true_mean

    # Perform MCMC to predict the next point in the random walk
    iterations = num_points * 10  # Increase iterations
    proposal_sd = 0.1  # Refine the proposal standard deviation
    predictions = metropolis_hastings_mean_random_walk(iterations, proposal_sd, true_walk)

    # Add the predictions to the true random walk to generate a new random walk
    new_walk = true_walk + predictions

    # Plot the true random walk along with the new random walk
    plt.figure(figsize=(10, 6))
    plt.plot(true_walk, label='True Random Walk', color='blue')
    plt.plot(new_walk, label='New Random Walk', linestyle='--', color='red')
    plt.xlabel('Step')
    plt.ylabel('Value')
    plt.title('Generating More Accurate Random Walk using MCMC Predictions')
    plt.legend()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from tensorflow.keras.optimizers import Adam

# Generate the random walk data
def random_walk_2d(num_steps):
    x = np.zeros(num_steps + 1)
    y = np.zeros(num_steps + 1)

    for i in range(1, num_steps + 1):
        # Generate random steps in both x and y directions
        step_x = np.random.choice([-1, 1])
        step_y = np.random.choice([-1, 1])

        # Update position
        x[i] = x[i - 1] + step_x
        y[i] = y[i - 1] + step_y

    return x, y

def calculate_distance(x, y):
    return np.sqrt(x**2 + y**2)

# Generate random walk data
num_steps = 1000
x, y = random_walk_2d(num_steps)
distances = calculate_distance(x, y)

# Prepare the data for the RNN model
X = np.column_stack((x, y))
y = distances

# Reshape data to fit the RNN input requirements (samples, timesteps, features)
X_rnn = X.reshape((X.shape[0], 1, X.shape[1]))

# Build the RNN model
model = Sequential()
model.add(SimpleRNN(50, activation='relu', input_shape=(1, 2)))
model.add(Dense(1))

# Compile the model
model.compile(optimizer=Adam(learning_rate=0.01), loss='mean_squared_error')

# Define iterative training process
num_iterations = 10
batch_size = 32
epochs_per_iteration = 10

for iteration in range(num_iterations):
    print(f"Iteration {iteration + 1}/{num_iterations}")
    # Train the RNN model iteratively
    model.fit(X_rnn, y, epochs=epochs_per_iteration, batch_size=batch_size, verbose=1)
    
    # Generate predictions after each iteration
    predictions = model.predict(X_rnn).flatten()
    
    # Calculate the shaded area
    shaded_area = np.sum(distances - predictions)
    print(f'Shaded area between actual and predicted curves: {shaded_area}')
    
    # Predict the next value
    last_data_point = np.array([x[-1], y[-1]]).reshape(1, 1, -1)  # Reshape to match RNN input shape
    next_prediction = model.predict(last_data_point)
    print('Next prediction:', next_prediction[0][0])

# Plot the actual data points and final predictions
plt.figure(figsize=(12, 6))

# Plot the 2D random walk
plt.subplot(1, 2, 1)
plt.plot(x, y, marker='o', linestyle='-', label='Actual Data')
plt.title('2D Random Walk')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal')  # Set aspect ratio to 'equal'
plt.legend()
plt.grid(True)

# Plot the distance from center vs time
plt.subplot(1, 2, 2)
plt.plot(range(len(distances)), distances, label='Actual Data')
plt.plot(range(len(predictions)), predictions, label='Predictions', linestyle='--')
plt.fill_between(range(len(distances)), distances, predictions, where=(distances >= predictions), color='skyblue', alpha=0.3)  # Fill area between curves where actual >= predictions
plt.fill_between(range(len(distances)), distances, predictions, where=(distances < predictions), color='salmon', alpha=0.3)  # Fill area between curves where actual < predictions
plt.title('Distance from Center vs Time')
plt.xlabel('Time')
plt.ylabel('Distance from Center')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def levy_flight_step(alpha, size=1):
    """
    Generate step sizes for a Lévy flight using a power-law distribution.
    """
    return np.random.pareto(alpha, size)

def levy_walk(n_steps, alpha, dimensions=1):
    """
    Perform a Lévy walk.
    """
    positions = np.zeros((n_steps, dimensions))
    for i in range(1, n_steps):
        direction = np.random.uniform(-1, 1, dimensions)
        direction /= np.linalg.norm(direction)
        step_size = levy_flight_step(alpha)
        positions[i] = positions[i - 1] + step_size * direction
    return positions

# Parameters
n_steps = 1000
alpha = 2.5
num_sims = 10000
dimensions = 1  # Set the number of dimensions for the walk

# Arrays to store the end positions
end_positions = []

# Plot the Lévy walks
plt.figure(figsize=(10, 6))

for i in range(num_sims):
    positions = levy_walk(n_steps, alpha, dimensions=dimensions)
    end_positions.append(positions[-1, 0])  # Collecting the end point in 1D
    
    if dimensions == 1:
        plt.plot(range(n_steps), positions[:, 0], alpha=0.6, linewidth=0.8)
    elif dimensions == 2:
        plt.plot(positions[:, 0], positions[:, 1], alpha=0.6, linewidth=0.8)

if dimensions == 1:
    plt.title(f"Lévy Walk Simulations ({num_sims} Runs, α={alpha}, 1D)")
    plt.xlabel("Step Index")
    plt.ylabel("Position")
elif dimensions == 2:
    plt.title(f"Lévy Walk Simulations ({num_sims} Runs, α={alpha}, 2D)")
    plt.xlabel("X")
    plt.ylabel("Y")

plt.grid(alpha=0.4)
plt.show()

# Histogram of the end positions
plt.figure(figsize=(8, 6))
plt.hist(end_positions, bins=90, density=True, color='skyblue', edgecolor='black', alpha=0.7)
plt.title("Histogram of End Positions (Lévy Walk)")
plt.xlabel("Final X Position")
plt.ylabel("Density")
plt.grid(alpha=0.4)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter, find_peaks

# Function to denoise stock data and compute its negative derivative
def denoise_stock_data_and_compute_derivative(csv_file, sensitivity):
    """
    Denoise stock closing price data using Savitzky-Golay filter and compute the negative derivative.

    Parameters:
    csv_file (str): Path to the CSV file containing 'Date' and 'Close' columns.
    sensitivity (int): Smoothing sensitivity (larger values smooth more, minimum 3).
    
    Returns:
    None: Plots the original data, denoised data, and the superimposed negative derivative with oscillation period.
    """
    # Read the CSV file
    data = pd.read_csv(csv_file)
    
    # Ensure 'Date' is datetime and sort data
    data['Date'] = pd.to_datetime(data['Date'])
    data.sort_values('Date', inplace=True)
    
    # Extract closing prices and dates
    close_prices = data['Close'].values
    dates = data['Date'].values
    
    # Denoise using Savitzky-Golay filter
    window_length = max(3, sensitivity)  # Window length must be >= 3 and odd
    if window_length % 2 == 0:  # Ensure window_length is odd
        window_length += 1
    
    polyorder = 1  # Degree of the polynomial to fit
    denoised_prices = savgol_filter(close_prices, window_length, polyorder)
    
    # Compute the negative derivative of the denoised data
    time_intervals = np.arange(len(denoised_prices))  # Time intervals as index positions
    negative_derivative = -np.gradient(denoised_prices, time_intervals)
    
    # Denoise the negative derivative
    denoised_negative_derivative = savgol_filter(negative_derivative, window_length, polyorder)
    
    # Find peaks and troughs in the denoised negative derivative
    peaks, _ = find_peaks(denoised_negative_derivative)  # Local maxima
    troughs, _ = find_peaks(-denoised_negative_derivative)  # Local minima (inverted negative derivative)
    
    # Calculate the periods (time between consecutive peaks)
    peak_times = dates[peaks]
    trough_times = dates[troughs]
    
    # Calculate the average period between peaks
    peak_periods = np.diff(peak_times).astype('timedelta64[D]').astype(int)  # Convert to days
    avg_peak_period = np.mean(peak_periods) if len(peak_periods) > 0 else None
    
    # Calculate the average period between troughs
    trough_periods = np.diff(trough_times).astype('timedelta64[D]').astype(int)  # Convert to days
    avg_trough_period = np.mean(trough_periods) if len(trough_periods) > 0 else None
    
    # Print the periods
    print(f"Average Period between Peaks: {avg_peak_period} days" if avg_peak_period is not None else "No peaks detected")
    print(f"Average Period between Troughs: {avg_trough_period} days" if avg_trough_period is not None else "No troughs detected")
    
    # Plot original and denoised data
    plt.figure(figsize=(12, 6))
    plt.subplot(2, 1, 1)  # First subplot for original vs denoised
    plt.plot(dates, close_prices, label='Original Close Prices', color='blue', alpha=0.6)
    plt.plot(dates, denoised_prices, label='Denoised Close Prices', color='red', linewidth=2)
    plt.title('Stock Closing Prices - Original vs Denoised')
    plt.xlabel('Date')
    plt.ylabel('Close Price')
    plt.legend()
    plt.grid(True)
    
    # Superimpose original and denoised negative derivative plots
    plt.subplot(2, 1, 2)  # Second subplot for the superimposed negative derivative
    plt.plot(dates, negative_derivative, label='Original Negative Derivative', color='blue', linestyle='dashed')
    plt.plot(dates, denoised_negative_derivative, label='Denoised Negative Derivative', color='green', linewidth=2)
    plt.scatter(peak_times, denoised_negative_derivative[peaks], color='red', label='Peaks', zorder=5)
    plt.scatter(trough_times, denoised_negative_derivative[troughs], color='purple', label='Troughs', zorder=5)
    plt.title('Negative Derivative of Denoised Close Prices')
    plt.xlabel('Date')
    plt.ylabel('Negative Derivative')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()

# Example usage
# Replace 'your_stock_data.csv' with your CSV file path
csv_file = 'C:\\Users\\vedan\\OneDrive\\Desktop\\pypractice\\0P0000XW4J.BO_quant_data.csv'
sensitivity = 15  # Adjust sensitivity for smoothing (e.g., 5, 15, 25)
denoise_stock_data_and_compute_derivative(csv_file, sensitivity)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
from scipy.fftpack import fft

# Function to process and analyze multiple CSV files
def process_and_analyze_multiple_files(csv_files, sensitivity):
    """
    Process multiple stock data CSV files, superimpose results on 4 plots, 
    and compute the covariance matrix of closing prices.

    Parameters:
    csv_files (list of str): List of paths to CSV files containing 'Date' and 'Close' columns.
    sensitivity (int): Smoothing sensitivity (larger values smooth more, minimum 3).
    
    Returns:
    None: Displays superimposed plots and covariance matrix.
    """
    # Store closing prices for covariance matrix
    all_closing_prices = []

    # Prepare plots
    plt.figure(figsize=(14, 20))

    for csv_file in csv_files:
        # Read the CSV file
        data = pd.read_csv(csv_file)
        data['Date'] = pd.to_datetime(data['Date'])
        data.sort_values('Date', inplace=True)

        # Extract closing prices and dates
        close_prices = data['Close'].values
        dates = data['Date'].values
        all_closing_prices.append(close_prices)

        # Denoise using Savitzky-Golay filter
        window_length = max(3, sensitivity)
        if window_length % 2 == 0:
            window_length += 1
        polyorder = 1
        denoised_prices = savgol_filter(close_prices, window_length, polyorder)

        # Compute derivatives
        time_intervals = np.arange(len(denoised_prices))
        negative_derivative = -np.gradient(denoised_prices, time_intervals)
        denoised_negative_derivative = savgol_filter(negative_derivative, window_length, polyorder)
        second_negative_derivative = -np.gradient(denoised_negative_derivative, time_intervals)
        denoised_second_negative_derivative = savgol_filter(second_negative_derivative, window_length, polyorder)

        # Fourier decomposition
        N = len(denoised_second_negative_derivative)
        freqs = np.fft.fftfreq(N, d=1)
        fft_values = fft(denoised_second_negative_derivative)
        magnitudes = np.abs(fft_values)[:N // 2]
        positive_freqs = freqs[:N // 2]

        # Superimpose results on the plots
        # Original vs Denoised
        plt.subplot(5, 1, 1)
        plt.plot(dates, close_prices, label=f'Original - {csv_file}', alpha=0.5)
        plt.plot(dates, denoised_prices, label=f'Denoised - {csv_file}', linewidth=1)

        # Negative Derivative
        plt.subplot(5, 1, 2)
        plt.plot(dates, denoised_negative_derivative, label=f'Derivative - {csv_file}', linewidth=1)

        # Second Negative Derivative
        plt.subplot(5, 1, 3)
        plt.plot(dates, denoised_second_negative_derivative, label=f'Second Derivative - {csv_file}', linewidth=1)

        # Fourier Decomposition
        plt.subplot(5, 1, 4)
        plt.plot(positive_freqs, magnitudes, label=f'FFT - {csv_file}', linewidth=1)

    # Finalize plots
    for i in range(1, 5):
        plt.subplot(5, 1, i)
        plt.legend()
        plt.grid(True)

    # Compute and plot covariance matrix
    all_closing_prices_df = pd.DataFrame(all_closing_prices).T
    cov_matrix = all_closing_prices_df.cov()
    plt.subplot(5, 1, 5)
    plt.imshow(cov_matrix, cmap='RdBu', interpolation='nearest')
    plt.title('Covariance Matrix of Closing Prices')
    plt.colorbar(label='Covariance')
    plt.xlabel('File Index')
    plt.ylabel('File Index')

    plt.tight_layout()
    plt.show()


# Example usage
csv_files = [
    '0P0000XW4J.BO_quant_data.csv',
    '0P0000XW7X.BO_hdfc_inf_data.csv',
    '0P00012ALS.BO_motilal_data.csv'
]
sensitivity = 15
process_and_analyze_multiple_files(csv_files, sensitivity)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_stock_prices(S0, x0, gamma, mu, sigma, k_min, k_max, T, dt):
    """
    Simulate stock prices using Cauchy-distributed returns with drift.
    
    Parameters:
    S0 (float): Initial stock price
    x0 (float): Location parameter of Cauchy distribution
    gamma (float): Scale parameter of Cauchy distribution
    mu (float): Drift parameter
    sigma (float): Volatility scaling for Cauchy component
    k_min (float): Minimum return cutoff
    k_max (float): Maximum return cutoff
    T (float): Total time duration
    dt (float): Time step
    
    Returns:
    np.array: Simulated stock prices
    """
    N = int(T / dt)  # Number of time steps
    prices = np.zeros(N + 1)
    prices[0] = S0
    
    for i in range(1, N + 1):
        # Generate return with drift and Cauchy noise
        r = mu * dt + sigma * (np.random.standard_cauchy() * gamma + x0)
        
        # Apply cutoff limits
        r = np.clip(r, k_min, k_max)
        
        # Update stock price
        prices[i] = prices[i - 1] * (1 + r)
    
    return prices

# Parameters
S0 = 100          # Initial stock price
x0 = 0.0         # Location parameter
gamma = 0.02     # Scale parameter
mu = 0.05        # Drift parameter
sigma = 1.0      # Volatility scaling for Cauchy component
k_min = -0.05    # Minimum return cutoff (-5%)
k_max = 0.05     # Maximum return cutoff (5%)
T = 1.0          # Total time duration (1 year)
dt = 1/252       # Time step (1 trading day)
num_simulations = 100000  # Number of simulations

# Generate multiple stock price evolutions
final_prices = np.zeros(num_simulations)
for i in range(num_simulations):
    stock_prices = generate_stock_prices(S0, x0, gamma, mu, sigma, k_min, k_max, T, dt)
    final_prices[i] = stock_prices[-1]

# Plot a single realization
plt.figure(figsize=(10, 5))
plt.plot(stock_prices, label='Stock Price')
plt.xlabel('Time Step')
plt.ylabel('Stock Price')
plt.title('Stock Price Evolution with Drift and Cauchy-Distributed Returns')
plt.legend()
plt.show()

# Plot histogram of final prices
plt.figure(figsize=(10, 5))
bins = 50
counts, bin_edges, _ = plt.hist(final_prices, bins=bins, alpha=0.75, edgecolor='black')
plt.xlabel('Final Stock Price')
plt.ylabel('Frequency')
plt.title('Histogram of Final Stock Prices After 1 Year')
plt.show()

# Compute area fractions
below_S0 = np.sum(final_prices < S0) / num_simulations
above_S0 = np.sum(final_prices > S0) / num_simulations

# Compute mode (price range with maximum frequency)
max_freq_idx = np.argmax(counts)
mode_range = (bin_edges[max_freq_idx], bin_edges[max_freq_idx + 1])

# Display area fractions and mode
plt.figure(figsize=(10, 5))
plt.hist(final_prices, bins=bins, alpha=0.75, edgecolor='black')
plt.axvline(S0, color='r', linestyle='dashed', linewidth=2, label=f'S0 = {S0}')
plt.axvspan(mode_range[0], mode_range[1], color='yellow', alpha=0.5, label=f'Mode Range: {mode_range[0]:.2f} - {mode_range[1]:.2f}')
plt.xlabel('Final Stock Price')
plt.ylabel('Frequency')
plt.title('Histogram of Final Stock Prices with Area Fractions')
plt.legend()
plt.text(S0 * 1.1, plt.ylim()[1] * 0.8, f'Below S0: {below_S0:.2%}', color='r')
plt.text(S0 * 1.1, plt.ylim()[1] * 0.7, f'Above S0: {above_S0:.2%}', color='b')
plt.show()
```
